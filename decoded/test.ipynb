```python
import numpy as np
from sklearn.metrics import accuracy_score, f1_score

# load Step 3 file
data = np.load("windows/P32_window_probs_ema.npz", allow_pickle=True)

y_true = data["y_true"]
y_pred_raw = data["y_pred_raw"]
y_pred_smooth = data["y_pred_smooth"]

# safety: remove empty case
mask = y_true.size > 0
if not mask:
    print("No ground truth labels available.")
else:
    acc_raw = accuracy_score(y_true, y_pred_raw)
    acc_smooth = accuracy_score(y_true, y_pred_smooth)

    f1_raw = f1_score(y_true, y_pred_raw, average="macro")
    f1_smooth = f1_score(y_true, y_pred_smooth, average="macro")

    print(f"Raw Accuracy:     {acc_raw:.4f}")
    print(f"Smoothed Accuracy:{acc_smooth:.4f}")
    print()
    print(f"Raw Macro F1:     {f1_raw:.4f}")
    print(f"Smoothed Macro F1:{f1_smooth:.4f}")
```

```python
# where does smoothing change predictions?
changed = (y_pred_raw != y_pred_smooth)

print("Fraction changed:", changed.mean())

# are most changes near raw errors?
raw_correct = (y_pred_raw == y_true)

print("Changed where raw was correct:",
      np.mean(changed & raw_correct))

print("Changed where raw was wrong:",
      np.mean(changed & (~raw_correct)))
```
